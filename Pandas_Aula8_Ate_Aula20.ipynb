## Estatísticas descritivas

#### A estatística descritiva é a área da estatística que busca resumir, organizar e descrever os dados de forma clara e concisa. Ela nos permite obter uma visão geral dos dados e identificar padrões e tendências importantes. Como analista, isso é essencial para tomar decisões informadas e comunicar insights de maneira eficaz.

## Medidas de tendência central

### 1 - Média Aritmética: 
   * Definição e cálculo da média aritmética: A média aritmética é a medida de tendência central mais comumente utilizada. Ela é obtida somando todos os valores de um conjunto de dados e dividindo pelo número total de observações. Por exemplo, se tivermos os valores [3, 5, 7, 11, 15], a média seria (3 + 5 + 7 + 11 + 15) / 5 = 8.2.
   
#### OBS: Ela é sensível a valores extremos, pois cada valor contribui igualmente para o cálculo.

```python
salarios = [10000, 8000, 12000, 9000, 13000, 50000]
salarios = pd.Series(salarios)
salarios.mean()
```

### 2 - Mediana

   * Definição e cálculo da mediana: A mediana é o valor que divide o conjunto de dados em duas partes iguais. Para calcular a mediana, é necessário ordenar os valores em ordem crescente e selecionar o valor do meio. Por exemplo, para o conjunto [3, 5, 7, 11, 15], a mediana seria 7.
   * Comparação entre a mediana e a média: A mediana é uma medida de tendência central mais robusta à presença de valores extremos do que a média. Ela é menos influenciada por valores discrepantes e fornece uma estimativa melhor do centro dos dados em algumas situações.

```python
salarios.median()
```

### 3 - Moda

   * Definição e identificação da moda: A moda é o valor mais frequente em um conjunto de dados. Em um conjunto unimodal, existe apenas uma moda, enquanto em um conjunto multimodal, há várias modas. Por exemplo, em [3, 5, 5, 7, 11, 15], a moda é 5.
   * Casos em que os dados podem ter uma ou várias modas: A moda é especialmente útil quando lidamos com variáveis categóricas ou quando queremos identificar os valores mais comuns em um conjunto de dados.
   * Utilização da moda em distribuições de dados categóricos: A moda é amplamente utilizada em áreas como estudos de opinião, marketing e pesquisa de mercado para identificar as categorias mais frequentes.

```python
salarios = [10000, 8000, 12000, 9000, 9000, 9000, 13000, 9000, 10000]
salarios = pd.Series(salarios)
salarios.mode()
```

## Medidas de dispersão

### Essas medidas nos ajudam a compreender a variabilidade dos dados e fornecem informações importantes sobre a distribuição dos valores. 

### 1 - Desvio Padrão

   * O desvio padrão é uma medida que indica o quanto os valores de um conjunto de dados se afastam da média.
   * Ele nos permite compreender a dispersão ou variabilidade dos dados em relação à média.
   * Quanto maior o desvio padrão, maior é a dispersão dos dados em relação à média. Isso indica que os valores estão mais espalhados e há maior variabilidade.
   * Por outro lado, um desvio padrão baixo indica que os valores estão mais próximos da média e há menor variabilidade nos dados.
   * No exemplo dos retornos diários da ação, um desvio padrão alto sugere que os retornos variam bastante, o que pode indicar um investimento mais arriscado. Já um desvio padrão baixo indica que os retornos são mais estáveis.

```python
print(salarios.mean())
print(salarios.std())
```

#### * Dispersão dos salários: O desvio padrão de 1615 indica a dispersão ou variabilidade dos salários em relação à média. Valores mais altos de desvio padrão indicariam uma dispersão maior, enquanto um desvio padrão menor indica que os salários estão relativamente próximos da média.
#### * Variação dos salários: O desvio padrão de 1615 sugere que a maioria dos salários está dentro de uma faixa de aproximadamente 1615 pra cima ou pra baixo da média de 9888. Isso nos dá uma ideia da variação esperada dos salários na base de dados.

### Variância 

#### A variância é uma medida de dispersão que nos ajuda a entender o quão espalhados estão os dados em relação à média. Imagine que você está analisando a variação dos preços de ações em um determinado período de tempo.

* A variância é uma medida que quantifica o quão distantes os valores de um conjunto de dados estão da média.
* A variância é uma medida de dispersão que nos indica quão distantes os valores estão da média.
* Quanto maior a variância, maior é a dispersão dos dados em relação à média.
* Uma variância baixa indica que os valores estão mais próximos da média, sugerindo menor dispersão ou menor variabilidade nos dados.

```python
print(salarios.mean())
print(salarios.std())
print(salarios.var())
```

O resultado da variância é uma medida de dispersão que indica o quão distantes os valores estão da média, mas expresso em termos quadrados. Neste caso, a variância é de 2611111.111111111.

Porém, é importante notar que a variância é uma medida de dispersão ao quadrado, o que torna a interpretação direta do valor menos intuitiva. Para obter uma medida de dispersão na mesma escala dos dados originais, você pode utilizar o desvio padrão, que é a raiz quadrada da variância.

### Amplitude

#### uma medida de dispersão simples, porém importante. A amplitude nos ajuda a entender a extensão total dos dados, fornecendo informações sobre a diferença entre o maior e o menor valor em um conjunto de dados. 

* A amplitude é uma medida simples de dispersão que representa a diferença entre o maior e o menor valor em um conjunto de dados.
* Para calcular a amplitude, basta subtrair o menor valor do maior valor.
* Por exemplo, considere os seguintes dados de altura (em centímetros): [160, 165, 170, 175, 180]. A amplitude seria 180 - 160 = 20.
* A amplitude nos fornece uma noção básica da variação total dos dados.
* Quanto maior a amplitude, maior é a diferença entre os valores extremos do conjunto de dados.
* Uma amplitude baixa indica que os valores estão mais próximos uns dos outros, sugerindo uma variação menor nos dados.
* Ela é sensível a valores extremos, o que pode distorcer a interpretação se houver valores discrepantes.

```python
amplitude = salarios.max() - salarios.min()
amplitude
```

No entanto, é importante lembrar que a amplitude é uma medida simples de dispersão que não leva em consideração a distribuição dos salários entre o maior e o menor valor. Portanto, a amplitude por si só não fornece informações sobre a variação dos salários dentro dessa faixa. Pode haver uma grande variação entre os salários intermediários ou uma distribuição relativamente uniforme.

A interpretação mais adequada da amplitude depende do contexto específico dos dados e do domínio em questão. No caso de salários, a amplitude pode indicar a diferença de renda entre o funcionário com o salário mais alto e o funcionário com o salário mais baixo, mas não nos dá informações detalhadas sobre a distribuição dos salários intermediários.

### Quartis e intervalo interquartil

####  duas medidas de dispersão que nos ajudam a entender a distribuição dos dados de forma mais detalhada.

* Os quartis dividem um conjunto de dados ordenados em quatro partes iguais.
* O primeiro quartil (Q1) é o valor abaixo do qual 25% dos dados estão localizados.
* O segundo quartil (Q2), que é equivalente à mediana, divide o conjunto de dados em duas partes iguais: 50% dos dados estão abaixo dele e 50% estão acima.
* O terceiro quartil (Q3) é o valor abaixo do qual 75% dos dados estão localizados.
* Por exemplo, considere as seguintes notas em um exame: [60, 65, 70, 75, 80, 85]. Os quartis seriam Q1 = 65, Q2 = 72.5 e Q3 = 80.

### Intervalo interquartil:

* O intervalo interquartil é a diferença entre o terceiro quartil (Q3) e o primeiro quartil (Q1).
* Ele mede a dispersão dos dados em torno da mediana, ignorando os valores extremos.
* O intervalo interquartil fornece uma medida robusta de dispersão, menos sensível a valores extremos.
* Por exemplo, no conjunto de notas [60, 65, 70, 75, 80, 85], o intervalo interquartil seria 80 - 65 = 15.

```python
# Calcular os quartis 
q1 = salarios.quantile(0.25)
q2 = salarios.quantile(0.5)
q3 = salarios.quantile(0.75)

# Calcular o intervalo interquartil
iqr = q3 - q1

# Imprimir os resultados
print("Primeiro Quartil (Q1):", q1)
print("Segundo Quartil (Q2):", q2)
print("Terceiro Quartil (Q3):", q3)
print("Intervalo Interquartil (IQR):", iqr)
```
